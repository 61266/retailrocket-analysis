```python
import matplotlib.pyplot as plt

# Данные нашей воронки
stages = ['Просмотр товара', 'Добавление в корзину', 'Покупка']
users = [1404179, 32272, 9682]

# Создаём график
plt.figure(figsize=(10, 6))
bars = plt.bar(stages, users)

# Подписываем значения
for bar in bars:
    value = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value,
        f'{int(value):,}'.replace(',', ' '),
        ha='center',
        va='bottom',
        fontsize=11
    )

plt.title('Воронка интернет-магазина RetailRocket', fontsize=14)
plt.ylabel('Количество пользователей')
plt.xlabel('Этап воронки')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.savefig('retailrocket_funnel.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
view_to_cart = users[1] / users[0] * 100
cart_to_purchase = users[2] / users[1] * 100
view_to_purchase = users[2] / users[0] * 100

print(f'Конверсия из просмотра в корзину: {view_to_cart:.2f}%')
print(f'Конверсия из корзины в покупку: {cart_to_purchase:.2f}%')
print(f'Общая конверсия из просмотра в покупку: {view_to_purchase:.2f}%')
```
